## 1. Prawda, fałsz i "truthy / falsy"

W module 1 poznałeś/aś typ `bool` (`True` / `False`) i operatory porównania. To właśnie
wyrażenia logiczne będą warunkiem sterującym `if`-ami i pętlami.

Ważna rzecz, która ułatwia życie: `if` nie wymaga *dosłownie* wartości `bool` — każda
wartość w Pythonie da się ocenić jako "prawdziwa" (**truthy**) albo "fałszywa"
(**falsy**).

````python
# Wartości, które Python traktuje jako "fałszywe" (falsy):
print(bool(0))       # False
print(bool(0.0))     # False
print(bool(""))      # False — pusty string
print(bool(None))    # False
print(bool([]))      # False — pusta lista (poznamy szerzej w module 3)

# Prawie wszystko inne jest "prawdziwe" (truthy):
print(bool(1))
print(bool(-5))
print(bool("0"))     # UWAGA: to string "0", niepusty -> True!
print(bool("tekst"))
````

<div style="background-color:#eef6ff;border-left:5px solid #2b6cb0;padding:10px 15px;margin:12px 0;border-radius:4px;color:#1a365d;">
<b>💡 Ciekawostka</b><br>
Wyrażenie <code>bool(&quot;0&quot;)</code> daje <code>True</code>, bo to niepusty string (fakt, że w środku jest znak &quot;0&quot;, nie ma znaczenia). To klasyczna pułapka przy walidacji danych z <code>input()</code> — pusty string jest falsy, ale string zawierający cyfrę &quot;0&quot; już nie.
</div>

## 2. Instrukcja `if` / `elif` / `else`

Podstawowa składnia:

```python
if warunek:
    # kod, gdy warunek jest prawdziwy
elif inny_warunek:
    # kod, gdy pierwszy warunek fałszywy, a ten prawdziwy
else:
    # kod, gdy żaden z powyższych warunków nie jest prawdziwy
```

`elif` i `else` są opcjonalne — możesz mieć samo `if`, `if`/`else`, albo dowolną liczbę
`elif` między nimi. Warunki sprawdzane są **po kolei, od góry**, i wykonuje się blok
tylko dla pierwszego warunku, który okaże się prawdziwy — reszta jest pomijana.

````python
wiek = 20

if wiek < 13:
    print("Dziecko")
elif wiek < 18:
    print("Nastolatek")
elif wiek < 65:
    print("Dorosły")
else:
    print("Senior")
````

<b>⚠️ Częsty błąd</b><br>
Pojedynczy znak <code>=</code> to przypisanie (<code>x = 5</code>), a podwójny <code>==</code> to porównanie (<code>x == 5</code>). Pomylenie ich w <code>if</code> w Pythonie od razu zwróci błąd składni — w przeciwieństwie do niektórych innych języków, gdzie taki błąd potrafi się długo ukrywać.
</div>

## 3. Wcięcia — to nie kosmetyka

To jedna z najbardziej charakterystycznych cech Pythona: **wcięcie (indentation) jest
częścią składni**, a nie tylko stylem. Blok kodu należący do `if`, `while`, `for`, funkcji
itd. musi być konsekwentnie wcięty (zwykle 4 spacje).

````python
# Poprawnie:
if True:
    print("To jest w bloku if")
print("To już nie jest w bloku if")

# Poniższe (odkomentowane) wywoła IndentationError, bo brakuje wcięcia:
# if True:
# print("błąd składni")
````

<div style="background-color:#fff5f5;border-left:5px solid #e53e3e;padding:10px 15px;margin:12px 0;border-radius:4px;color:#742a2a;">
<b>⚠️ Uwaga na mieszanie spacji i tabulatorów</b><br>
Python 3 nie pozwala mieszać spacji i tabulatorów w tym samym bloku — zgłosi błąd <code>TabError</code>. W Jupyterze klawisz Tab zwykle sam wstawia 4 spacje, więc w praktyce rzadko to problem, ale warto wiedzieć, skąd może się wziąć dziwny błąd przy kopiowaniu kodu z różnych źródeł.
</div>

## 4. Zagnieżdżone warunki i operator warunkowy

`if` może zawierać kolejny `if` w środku (zagnieżdżenie) — przydatne, gdy druga decyzja
zależy od tego, że pierwszy warunek już jest spełniony. Dla prostych przypadków "wartość
A albo B w zależności od warunku" jest też krótszy zapis — **operator warunkowy (ternary)**:
`wartosc_a if warunek else wartosc_b`.

````python
temperatura = 5
pada_deszcz = True

if temperatura < 0:
    print("Ubierz się ciepło")
    if pada_deszcz:
        print("...i weź parasol, choć w mróz to rzadkość")
else:
    print("Nie jest zimno")

# To samo co prosty if/else, zapisane w jednej linii:
status = "ciepło" if temperatura > 15 else "zimno"
print(status)
````

## 5. Pętla `while`

Pętla `while` wykonuje blok kodu **dopóki warunek jest prawdziwy** — sprawdza warunek
przed każdym powtórzeniem. To Ty odpowiadasz za to, żeby warunek w końcu stał się fałszywy
(inaczej dostaniesz nieskończoną pętlę).

````python
licznik = 1
while licznik <= 5:
    print(f"Przebieg numer {licznik}")
    licznik += 1   # to samo co: licznik = licznik + 1

print("Koniec pętli")
````

<div style="background-color:#fff5f5;border-left:5px solid #e53e3e;padding:10px 15px;margin:12px 0;border-radius:4px;color:#742a2a;">
<b>⚠️ Nieskończona pętla</b><br>
Jeśli zapomnisz zaktualizować zmienną sterującą (np. <code>licznik += 1</code>), warunek nigdy się nie zmieni i pętla będzie działać w nieskończoność, blokując kernel Jupytera. W takiej sytuacji zatrzymujesz wykonanie przyciskiem <b>■ (Stop)</b> w pasku narzędzi albo z menu <i>Kernel → Interrupt</i>.
</div>

<div style="background-color:#eef6ff;border-left:5px solid #2b6cb0;padding:10px 15px;margin:12px 0;border-radius:4px;color:#1a365d;">
<b>💡 Wzorzec: while True + break</b><br>
Bardzo częsty schemat to <code>while True:</code> (warunek zawsze prawdziwy) z warunkiem wyjścia w środku pętli przez <code>break</code> — przydatne np. przy walidacji danych od użytkownika, gdzie chcemy pytać &quot;w kółko, aż odpowiedź będzie sensowna&quot;. Zobaczysz to w ćwiczeniach.
</div>

<a id="sec6"></a>
## 6. `break` i `continue`

Dwie instrukcje do sterowania przebiegiem pętli od środka:

- **`break`** — natychmiast **przerywa całą pętlę** (wychodzi z niej całkowicie).
- **`continue`** — **pomija resztę bieżącego przebiegu** i przechodzi do następnego.

````python
# break — szukamy pierwszej liczby podzielnej przez 7 w zakresie 1-100
n = 1
while n <= 100:
    if n % 7 == 0:
        print(f"Znaleziono: {n}")
        break
    n += 1

print("---")

# continue — wypisz liczby 1-10, ale pomiń parzyste
i = 0
while i < 10:
    i += 1
    if i % 2 == 0:
        continue
    print(i)
````

## 7. Pętla `for` i `range()`

Pętla `for` iteruje po **sekwencji** — najczęściej po wyniku funkcji `range()`, która
generuje kolejne liczby całkowite. `range()` ma trzy warianty wywołania:

| Wywołanie | Co generuje |
|---|---|
| `range(stop)` | `0, 1, 2, ..., stop-1` |
| `range(start, stop)` | `start, start+1, ..., stop-1` |
| `range(start, stop, krok)` | co `krok`, od `start` do `stop-1` (krok może być ujemny) |

Zauważ: **`stop` nigdy nie jest osiągnięty** — `range(5)` kończy się na 4.

````python
for i in range(5):
    print(i, end=" ")
print()

for i in range(2, 10, 2):
    print(i, end=" ")
print()

for i in range(10, 0, -1):     # liczenie wstecz
    print(i, end=" ")
print("\nStart!")
````

<div style="background-color:#eef6ff;border-left:5px solid #2b6cb0;padding:10px 15px;margin:12px 0;border-radius:4px;color:#1a365d;">
<b>💡 `for` kontra `while`</b><br>
Jeśli z góry wiesz <b>ile razy</b> coś ma się powtórzyć (albo iterujesz po znanej sekwencji) — użyj <code>for</code>. Jeśli liczba powtórzeń zależy od warunku, który poznasz dopiero w trakcie (np. &quot;dopóki użytkownik nie zgadnie&quot;) — użyj <code>while</code>. Każdą pętlę <code>for</code> da się zapisać jako <code>while</code>, ale rzadko warto.
</div>

## 8. Iterowanie po tekście (i zapowiedź list)

`for` działa na każdej sekwencji, nie tylko na `range()`. String też jest sekwencją znaków,
więc można po nim iterować bezpośrednio.

````python
for znak in "Python":
    print(znak)

# Policzmy, ile jest samogłosek w słowie:
slowo = "programowanie"
samogloski = "aeiouyą ę"  # (uproszczone, dla przykładu)
licznik_samoglosek = 0
for znak in slowo:
    if znak in samogloski:
        licznik_samoglosek += 1

print(f"Liczba samogłosek w '{slowo}': {licznik_samoglosek}")
````

<div style="background-color:#eef6ff;border-left:5px solid #2b6cb0;padding:10px 15px;margin:12px 0;border-radius:4px;color:#1a365d;">
<b>💡 Zapowiedź</b><br>
Operator <code>in</code> użyty wyżej (<code>znak in samogloski</code>) sprawdza przynależność — działa tak samo na stringach, jak i na **listach**, które poznamy szczegółowo w module 3 (<code>for element in lista:</code> wygląda identycznie jak iterowanie po stringu).
</div>

## 9. Zagnieżdżone pętle

Pętla może zawierać kolejną pętlę w środku — wtedy dla **każdego** przebiegu pętli
zewnętrznej wykonuje się **cała** pętla wewnętrzna. Klasyczny przykład: tabliczka mnożenia.

````python
for a in range(1, 4):
    for b in range(1, 4):
        print(f"{a} x {b} = {a * b}")
    print("---")   # separator po każdej "rundzie" pętli zewnętrznej
````

<div style="background-color:#fff5f5;border-left:5px solid #e53e3e;padding:10px 15px;margin:12px 0;border-radius:4px;color:#742a2a;">
<b>⚠️ Uważaj na złożoność</b><br>
Dwie zagnieżdżone pętle po 100 przebiegów każda to już 10 000 wykonań bloku wewnętrznego (100 × 100). Przy większych zakresach zagnieżdżone pętle potrafią zauważalnie spowolnić program — to temat, do którego wrócimy przy okazji złożoności obliczeniowej.
</div>

## 10. Ciekawostki: `for...else`, `pass`

Dwie rzadziej spotykane, ale warte poznania konstrukcje.

````python
# for...else: blok "else" wykonuje się, TYLKO jeśli pętla zakończyła się
# normalnie (bez break). Przydatne np. przy szukaniu czegoś.
liczby = [4, 8, 15, 16, 23, 42]
szukana = 99

for l in liczby:
    if l == szukana:
        print("Znaleziono!")
        break
else:
    print(f"{szukana} nie występuje w liście")

# pass: "nic nie rób" — placeholder, gdy składnia wymaga bloku,
# ale nie masz jeszcze co w nim napisać.
if True:
    pass  # tu kiedyś coś dopiszę
print("Program działa mimo pustego bloku if")
````

<div style="background-color:#eef6ff;border-left:5px solid #2b6cb0;padding:10px 15px;margin:12px 0;border-radius:4px;color:#1a365d;">
<b>💡 Ciekawostka</b><br>
<code>for...else</code> bywa mylące dla początkujących (i nie tylko) — słowo &quot;else&quot; sugeruje &quot;w przeciwnym razie&quot;, a tu oznacza raczej &quot;jeśli nie było <code>break</code>&quot;. Guido van Rossum przyznał kiedyś, że gdyby projektował to jeszcze raz, prawdopodobnie nazwałby tę konstrukcję inaczej.
</div>

<b>🔥 Ćwiczenie 8 (wyzwanie): Mini-gra: zgadnij liczbę</b><br>
Napisz grę: komputer losuje liczbę od 1 do 100 (<code>import random</code>, <code>random.randint(1, 100)</code>), a użytkownik zgaduje przez <code>input()</code>. Po każdej próbie wypisz &quot;za dużo&quot; albo &quot;za mało&quot;, aż użytkownik trafi — wtedy wypisz gratulacje i liczbę wykonanych prób. Użyj <code>while True</code> z <code>break</code> przy trafieniu.
</div>

<details>
<summary><b>👉 Kliknij, żeby zobaczyć przykładowe rozwiązanie</b></summary>

```python
import random

liczba_do_zgadniecia = random.randint(1, 100)
liczba_prob = 0

print("Zgadnij liczbę od 1 do 100!")

while True:
    proba = int(input("Twoja liczba: "))
    liczba_prob += 1

    if proba < liczba_do_zgadniecia:
        print("Za mało!")
    elif proba > liczba_do_zgadniecia:
        print("Za dużo!")
    else:
        print(f"Trafiłeś/aś! Zajęło Ci to {liczba_prob} prób.")
        break
```

Podpowiedź: to jest dokładnie ten wzorzec &quot;while True + break&quot; z sekcji 5 — pętla kręci się w nieskończoność, dopóki warunek trafienia (<code>proba == liczba_do_zgadniecia</code>) nie przerwie jej przez <code>break</code>.
