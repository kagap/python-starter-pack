## 3. Pierwszy program — `print()`

Funkcja `print()` wypisuje tekst (albo wynik działania) na ekranie. To najczęściej
używana funkcja do szybkiego sprawdzania "co się dzieje" w kodzie.

````python
print("Witaj, świecie!")
````

`print()` przyjmuje **wiele argumentów** naraz (rozdzielonych przecinkiem) oraz dwa
przydatne parametry:

- `sep` — czym rozdzielić kolejne argumenty (domyślnie spacja),
- `end` — co dodać na końcu (domyślnie znak nowej linii `\n`).

<div style="background-color:#eef6ff;border-left:5px solid #2b6cb0;padding:10px 15px;margin:12px 0;border-radius:4px;color:#1a365d;">
<b>💡 Ciekawostka</b><br>
W Pythonie 2 <code>print</code> był instrukcją (<code>print "tekst"</code>, bez nawiasów). W Pythonie 3 stał się zwykłą funkcją — to jedna z najbardziej widocznych różnic między tymi dwiema wersjami języka.
</div>

````python
print("t", "t", 2026, sep=" | ")
print("Ta linijka", end=" ")
print("kończy się bez przejścia do nowej linii, więc jest tutaj.")
````

## 4. Zmienne i typy danych

Zmienną tworzysz po prostu przez przypisanie wartości — **nie deklarujesz typu z góry**,
Python sam go rozpozna (tzw. *dynamiczne typowanie*).

Cztery podstawowe typy, od których zaczynamy:

| Typ | Przykład | Opis |
|---|---|---|---|
| `int` | `42` | liczba całkowita |
| `float` | `3.14` | liczba zmiennoprzecinkowa (z częścią dziesiętną) |
| `str` | `"tekst"` | ciąg znaków (string) |
| `bool` | `True` / `False` | wartość logiczna |

Typ aktualnej wartości zmiennej sprawdzisz funkcją `type()`.

````python
wiek = 30
pi = 3.14
imie = "k"
czy_studiuje = False

print(wiek, type(wiek))
print(pi, type(pi))
print(imie, type(imie))
print(czy_studiuje, type(czy_studiuje))
````

<div style="background-color:#fff5f5;border-left:5px solid #e53e3e;padding:10px 15px;margin:12px 0;border-radius:4px;color:#742a2a;">
<b>⚠️ Nazewnictwo zmiennych</b><br>
Nazwa zmiennej: zaczyna się od litery lub podkreślnika (nie od cyfry), może zawierać litery, cyfry i <code>_</code>, jest wielko-/małoliterowa (<code>wiek</code> i <code>Wiek</code> to dwie różne zmienne), i nie może być słowem kluczowym Pythona (np. <code>class</code>, <code>for</code>, <code>if</code>). Konwencja w Pythonie to <code>snake_case</code>, np. <code>liczba_zamowien</code>.
</div>

<div style="background-color:#eef6ff;border-left:5px solid #2b6cb0;padding:10px 15px;margin:12px 0;border-radius:4px;color:#1a365d;">
<b>💡 Ciekawostka</b><br>
Python stosuje tzw. <i>duck typing</i> — &laquo;jeśli coś chodzi jak kaczka i kwacze jak kaczka, to jest kaczką&raquo;. Liczy się to, co dana wartość <b>potrafi</b> zrobić, a nie jaki formalnie ma typ zadeklarowany. Dlatego ta sama zmienna może w trakcie działania programu &laquo;zmienić typ&raquo; — po prostu przypisując jej nową wartość.
</div>

## 5. Komentarze

Komentarz to fragment kodu, który Python **ignoruje** przy wykonywaniu — służy tylko
ludziom, żeby zrozumieli, co się dzieje (albo dlaczego). Krótkie komentarze zapisujesz po
`#`, dłuższe — w potrójnych cudzysłowach.

````python
# To jest komentarz jednoliniowy — Python go zignoruje
wynagrodzenie = 8500  # komentarz może też być na końcu linii z kodem

"""
To jest dłuższy komentarz (albo tzw. docstring),
przydatny np. do opisania, co robi cały blok kodu
albo funkcja.
"""
print(wynagrodzenie)
````

<a id="sec6"></a>
## 6. Operatory

**Arytmetyczne**

| Operator | Działanie | Przykład | Wynik |
|---|---|---|---|
| `+` | dodawanie | `5 + 2` | `7` |
| `-` | odejmowanie | `5 - 2` | `3` |
| `*` | mnożenie | `5 * 2` | `10` |
| `/` | dzielenie (zawsze zwraca `float`) | `5 / 2` | `2.5` |
| `//` | dzielenie całkowite | `5 // 2` | `2` |
| `%` | reszta z dzielenia (modulo) | `5 % 2` | `1` |
| `**` | potęgowanie | `5 ** 2` | `25` |

**Porównania** (zwracają `bool`): `==`, `!=`, `<`, `>`, `<=`, `>=`

**Logiczne**: `and`, `or`, `not`

````python
a = 17
b = 5

print(a + b, a - b, a * b)
print(a / b)    # zawsze float
print(a // b)   # ile pełnych "piątek" mieści się w 17
print(a % b)    # co zostaje z reszty

print(a > b, a == b, a != b)
print(a > 10 and b < 10)
print(a > 100 or b < 10)
````

<div style="background-color:#eef6ff;border-left:5px solid #2b6cb0;padding:10px 15px;margin:12px 0;border-radius:4px;color:#1a365d;">
<b>💡 Ciekawostka</b><br>
Kombinacja <code>//</code> i <code>%</code> to klasyczny sposób na rozbicie liczby na części, np. sekund na godziny/minuty/sekundy, albo sprawdzenie parzystości liczby (<code>n % 2 == 0</code>). Wrócimy do tego przy instrukcjach warunkowych w kolejnym module.
</div>

## 7. Konwersja typów i `input()`

Funkcja `input()` pobiera od użytkownika tekst wpisany w konsoli — i **zawsze** zwraca go
jako `str`, nawet jeśli osoba wpisze liczbę. Dlatego wynik trzeba świadomie
przekonwertować funkcją `int()` albo `float()`, jeśli chcemy na nim liczyć.

<div style="background-color:#fff5f5;border-left:5px solid #e53e3e;padding:10px 15px;margin:12px 0;border-radius:4px;color:#742a2a;">
<b>⚠️ Klasyczny błąd początkujących</b><br>
<code>input("Podaj wiek: ") + 1</code> wywali błąd (<code>TypeError</code>), bo próbujesz dodać liczbę do tekstu. Trzeba najpierw zamienić: <code>int(input("Podaj wiek: ")) + 1</code>.
</div>

````python
# Odkomentuj i uruchom, żeby przetestować interaktywnie:
# tekst = input("Podaj swój wiek: ")
# print(type(tekst))          # zawsze <class 'str'>
# wiek = int(tekst)
# print(type(wiek), wiek + 1) # teraz już można liczyć

# Konwersje w drugą stronę też działają:
liczba = 42
print(str(liczba) + " to teraz tekst")
print(float("3.5") + 1)
print(int("10") + int("5"))
````

## 8. Formatowanie tekstu — f-stringi

Najwygodniejszy sposób na wstawianie wartości zmiennych do tekstu to **f-string**:
poprzedzasz cudzysłów literą `f`, a zmienne wstawiasz w `{nawiasach klamrowych}`. Można
też od razu formatować liczby, np. zaokrąglać do konkretnej liczby miejsc po przecinku.

````python
imie = "Kamil"
saldo = 1234.5678

print(f"Cześć, {imie}!")
print(f"Twoje saldo to {saldo:.2f} zł")   # zaokrąglenie do 2 miejsc po przecinku
print(f"{imie} ma na koncie {saldo:,.2f} zł".replace(",", " "))
````

<div style="background-color:#eef6ff;border-left:5px solid #2b6cb0;padding:10px 15px;margin:12px 0;border-radius:4px;color:#1a365d;">
<b>💡 Ciekawostka</b><br>
F-stringi pojawiły się dopiero w Pythonie 3.6 (2016 r.). Wcześniej trzeba było używać <code>"Cześć, {}!".format(imie)</code> albo jeszcze starszego <code>"Cześć, %s!" % imie</code>. F-stringi są zwykle najszybsze i najbardziej czytelne, dlatego dziś to domyślny wybór.
</div>
