## 1. Po co funkcje — zasada DRY

Gdybyś w module 1 chciał/a trzy razy policzyć BMI dla różnych osób, musiałbyś/abyś
skopiować ten sam wzór trzy razy. To prosta droga do błędów — gdy trzeba coś poprawić,
łatwo zapomnieć o jednej z kopii. Stąd zasada **DRY** (*Don't Repeat Yourself* — nie
powtarzaj się): powtarzalną logikę zamykamy w funkcji i wywołujemy ją tyle razy, ile
trzeba.

<a id="sec2"></a>
## 2. `def`, parametry, `return`

Funkcję definiuje się słowem kluczowym `def`, nazwą, listą parametrów w nawiasach i
dwukropkiem. `return` kończy działanie funkcji i zwraca wartość do miejsca, z którego ją
wywołano — funkcja bez `return` (albo z samym `return` bez wartości) zwraca `None`.

````python
def przywitaj(imie):
    return f"Cześć, {imie}!"

wiadomosc = przywitaj("Kamil")
print(wiadomosc)
print(przywitaj("Ania"))   # tę samą funkcję można wywołać dowolną liczbę razy


def oblicz_bmi(waga, wzrost):
    return waga / wzrost ** 2

print(oblicz_bmi(78, 1.80))
print(oblicz_bmi(65, 1.65))
````

> ⚠️ **`return` kontra `print`**
>
> To częste źródło pomyłek u początkujących: `print()` tylko **wyświetla** wartość na ekranie, natomiast `return` **zwraca** ją do dalszego użycia w programie (np. przypisania do zmiennej). Funkcja, która tylko `print()`-uje wynik, nie pozwala go potem wykorzystać w obliczeniach — `wynik = oblicz_bmi(...)` zadziała tylko, jeśli funkcja faktycznie użyje `return`.

<a id="sec3"></a>
## 3. Domyślne wartości parametrów

Parametr może mieć wartość domyślną, używaną, gdy wywołujący jej nie poda. Dzięki temu
funkcja bywa elastyczna, ale ma sensowne zachowanie "z automatu".

````python
def przywitaj(imie, powitanie="Cześć"):
    return f"{powitanie}, {imie}!"

print(przywitaj("Kamil"))                 # użyje domyślnego "Cześć"
print(przywitaj("Kamil", "Witaj"))        # nadpisze domyślną wartość
print(przywitaj("Kamil", powitanie="Hej"))  # to samo, ale jawnie nazwanym argumentem
````

<a id="sec4"></a>
## 4. Argumenty pozycyjne i nazwane

Argumenty można podawać **pozycyjnie** (w kolejności parametrów) albo **przez nazwę**
(`parametr=wartość`, wtedy kolejność nie ma znaczenia). Można je też mieszać, ale
argumenty pozycyjne zawsze muszą iść przed nazwanymi.

````python
def opisz_produkt(nazwa, cena, waluta="zł"):
    return f"{nazwa}: {cena} {waluta}"

print(opisz_produkt("Chleb", 4.5))
print(opisz_produkt(cena=12.0, nazwa="Ser"))          # kolejność nazwanych nie ma znaczenia
print(opisz_produkt("Kawa", 15.0, waluta="EUR"))

# To wywoła błąd - argument pozycyjny po nazwanym:
# opisz_produkt(nazwa="Kawa", 15.0)
````

<a id="sec5"></a>
## 5. `*args` i `**kwargs`

Czasem nie wiadomo z góry, ile argumentów funkcja dostanie. `*args` zbiera dowolną
liczbę argumentów pozycyjnych do krotki, a `**kwargs` — dowolną liczbę argumentów
nazwanych do słownika. Nazwy `args`/`kwargs` to tylko konwencja — liczy się `*` i `**`.

````python
def zsumuj_wszystko(*liczby):
    print(liczby, type(liczby))   # to krotka
    return sum(liczby)

print(zsumuj_wszystko(1, 2, 3))
print(zsumuj_wszystko(1, 2, 3, 4, 5))   # dowolna liczba argumentów


def wypisz_profil(**dane):
    print(dane, type(dane))       # to słownik
    for klucz, wartosc in dane.items():
        print(f"{klucz}: {wartosc}")

wypisz_profil(imie="Kamil", wiek=30, miasto="Warszawa")
````

<a id="sec6"></a>
## 6. Zasięg zmiennych (scope)

Zmienna zdefiniowana **wewnątrz** funkcji jest lokalna — istnieje tylko podczas jej
wykonania i jest niewidoczna na zewnątrz. Zmienna zdefiniowana **na zewnątrz** funkcji
(globalna) jest widoczna wewnątrz funkcji do odczytu, ale żeby ją **zmienić** wewnątrz
funkcji, trzeba jawnie użyć słowa kluczowego `global`.

````python
licznik_odwiedzin = 0   # zmienna globalna

def zwieksz_bez_global():
    licznik_odwiedzin = 100   # to tworzy NOWĄ zmienną LOKALNĄ, nie zmienia globalnej!
    print(f"Wewnątrz funkcji: {licznik_odwiedzin}")

def zwieksz_z_global():
    global licznik_odwiedzin
    licznik_odwiedzin += 1

zwieksz_bez_global()
print(f"Po pierwszej funkcji: {licznik_odwiedzin}")   # bez zmian, wciąż 0

zwieksz_z_global()
zwieksz_z_global()
print(f"Po dwóch wywołaniach z global: {licznik_odwiedzin}")   # 2
````

> ⚠️ **Unikaj nadużywania `global`**
>
> Modyfikowanie zmiennych globalnych wewnątrz funkcji utrudnia śledzenie, co i gdzie zmienia wartość - szczególnie w większych programach. Dobra praktyka: funkcja powinna dostawać dane przez parametry i zwracać wynik przez `return`, a `global` traktować jako wyjątek, nie normę.

<a id="sec7"></a>
## 7. Docstringi — dokumentowanie funkcji

Docstring to string dokumentujący umieszczony jako pierwsza linia ciała funkcji (w
potrójnych cudzysłowach) — opisuje, co funkcja robi, jakie przyjmuje argumenty i co
zwraca. Jest dostępny przez `help()` albo atrybut `.__doc__`.

````python
def oblicz_bmi(waga, wzrost):
    """
    Oblicza wskaźnik BMI.

    Argumenty:
        waga (float): waga w kilogramach
        wzrost (float): wzrost w metrach

    Zwraca:
        float: wartość BMI
    """
    return waga / wzrost ** 2

help(oblicz_bmi)
print(oblicz_bmi.__doc__)
````

<a id="sec8"></a>
## 8. Funkcje lambda

`lambda` tworzy małą, anonimową (bez nazwy) funkcję w jednej linii — przydatną tam, gdzie
funkcja jest potrzebna tylko na chwilę, np. jako argument innej funkcji (`sorted()`,
`key=`). Składnia: `lambda argumenty: wyrażenie` (bez `return` — wynik wyrażenia jest
zwracany automatycznie).

````python
kwadrat = lambda x: x ** 2
print(kwadrat(5))          # to samo co: def kwadrat(x): return x ** 2

# Typowe zastosowanie: sortowanie po niestandardowym kluczu
osoby = [("Ania", 25), ("Kamil", 30), ("Tomek", 20)]
posortowane_po_wieku = sorted(osoby, key=lambda osoba: osoba[1])
print(posortowane_po_wieku)
````

> 💡 **Kiedy używać lambdy**
>
> Lambda ma sens tylko dla bardzo krótkich, jednowyrazowych operacji. Jeśli logika robi się bardziej złożona (więcej niż jedna linijka, warunki), zwykła funkcja z `def` i sensowną nazwą będzie dużo czytelniejsza.

## 9. Rekurencja — krótkie wprowadzenie

Funkcja **rekurencyjna** wywołuje samą siebie, z coraz mniejszym/prostszym argumentem, aż
dojdzie do tzw. **przypadku bazowego**, który kończy rekurencję bez kolejnego wywołania.
Bez przypadku bazowego dostaniesz nieskończoną rekurencję (i błąd).

````python
def silnia(n):
    if n <= 1:            # przypadek bazowy - tu rekurencja się zatrzymuje
        return 1
    return n * silnia(n - 1)   # wywołanie samej siebie z mniejszym argumentem

print(silnia(5))    # 5 * 4 * 3 * 2 * 1 = 120

# To samo iteracyjnie, dla porównania:
def silnia_iteracyjna(n):
    wynik = 1
    for i in range(1, n + 1):
        wynik *= i
    return wynik

print(silnia_iteracyjna(5))
````

> 💡 **Ciekawostka**
>
> Python domyślnie ogranicza głębokość rekurencji do ok. 1000 wywołań (`sys.getrecursionlimit()`) - przekroczenie tego limitu wywoła `RecursionError`. To celowe zabezpieczenie przed brakującym przypadkiem bazowym, który inaczej zawiesiłby program w nieskończonej pętli wywołań.

> 📝 **Ćwiczenie 3: Min i max naraz**
>
> Napisz funkcję `min_i_max(lista)`, która zwraca **dwie** wartości naraz: najmniejszy i największy element listy (jako krotkę). Nie używaj wbudowanych `min()`/`max()` - policz to pętlą, żeby przećwiczyć logikę.

<details>
<summary><b>👉 Kliknij, żeby zobaczyć przykładowe rozwiązanie</b></summary>

```python
def min_i_max(lista):
    najmniejszy = lista[0]
    najwiekszy = lista[0]
    for element in lista:
        if element < najmniejszy:
            najmniejszy = element
        if element > najwiekszy:
            najwiekszy = element
    return najmniejszy, najwiekszy

mini, maxi = min_i_max([4, 8, 15, 16, 23, 42, 1])
print(f"Min: {mini}, Max: {maxi}")
```
</details>

> 📝 **Ćwiczenie 6: Zasięg zmiennych - znajdź błąd**
>
> Poniższy kod ma zwracać licznik zwiększony o 5, ale ktoś popełnił błąd zasięgu - znajdź go i popraw, żeby `print(wynik)` pokazał `15`.
>
> `licznik = 10`
>
> `def dodaj_piec():`
>
> &nbsp;&nbsp;&nbsp;&nbsp;`licznik = licznik + 5`
>
> &nbsp;&nbsp;&nbsp;&nbsp;`return licznik`
>
> `wynik = dodaj_piec()`

<details>
<summary><b>👉 Kliknij, żeby zobaczyć przykładowe rozwiązanie</b></summary>

```python
licznik = 10

def dodaj_piec(wartosc):
    return wartosc + 5

wynik = dodaj_piec(licznik)
print(wynik)   # 15
```

Podpowiedź / wyjaśnienie błędu: oryginalny kod próbował zmienić zmienną globalną `licznik` bez `global`, a dodatkowo `licznik = licznik + 5` wewnątrz funkcji próbuje odczytać `licznik`, zanim lokalna wersja zostanie przypisana - to daje `UnboundLocalError`. Najczystsze rozwiązanie to w ogóle nie używać zmiennej globalnej, tylko przekazać wartość jako argument, tak jak w rozwiązaniu powyżej.
</details>

> 🔥 **Ćwiczenie 8 (wyzwanie): Ciąg Fibonacciego rekurencyjnie**
>
> Napisz rekurencyjną funkcję `fibonacci(n)`, zwracającą n-ty wyraz ciągu Fibonacciego (0, 1, 1, 2, 3, 5, 8, 13, ...), gdzie `fibonacci(0) = 0`, `fibonacci(1) = 1`, a każdy kolejny to suma dwóch poprzednich. Wypisz wyrazy od 0 do 10. Zastanów się (albo sprawdź empirycznie), dlaczego to podejście robi się zauważalnie wolne dla większych `n`, np. `fibonacci(35)`.

<details>
<summary><b>👉 Kliknij, żeby zobaczyć przykładowe rozwiązanie</b></summary>

```python
def fibonacci(n):
    if n <= 1:              # przypadek bazowy
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

for i in range(11):
    print(fibonacci(i), end=" ")
```

Wyjaśnienie: ta wersja liczy te same wartości wielokrotnie (np. `fibonacci(3)` jest liczone od nowa za każdym razem, gdy jest potrzebne) - liczba wywołań rośnie wykładniczo wraz z `n`. Naprawia się to techniką zwaną memoizacją (zapamiętywaniem już policzonych wyników), do której wrócimy przy okazji bardziej zaawansowanych tematów.
</details>
