## 1. Po co nam kolekcje

Wyobraź sobie listę zakupów złożoną z 15 produktów. Robienie 15 osobnych zmiennych
(`produkt1`, `produkt2`, ...) byłoby absurdalne — i niemożliwe do obsłużenia pętlą, bo nie
da się napisać `for i in range(15): produkt{i}`. Dlatego Python ma wbudowane typy
**kolekcji**, które trzymają wiele wartości pod jedną nazwą zmiennej.

<a id="sec2"></a>
## 2. Listy — tworzenie, indeksowanie, wycinki

Lista (`list`) to uporządkowany, **mutowalny** (czyli można ją zmieniać po utworzeniu)
zbiór wartości, zapisywany w nawiasach kwadratowych. Może zawierać elementy różnych typów
naraz, choć w praktyce zwykle trzyma jeden typ.

Elementy numerujemy od **0**, a indeksy ujemne liczą od końca (`-1` to ostatni element).

````python
owoce = ["jabłko", "banan", "gruszka", "pomarańcza", "kiwi"]

print(owoce[0])       # pierwszy element
print(owoce[-1])      # ostatni element
print(owoce[1:3])     # wycinek: indeksy 1 i 2 (3 jest wyłączony)
print(owoce[:2])      # od początku do indeksu 2 (wyłącznie)
print(owoce[2:])      # od indeksu 2 do końca
print(len(owoce))     # liczba elementów
````

> ⚠️ **Uwaga na IndexError**
>
> Próba odwołania się do indeksu, którego nie ma (np. `owoce[10]` przy liście pięcioelementowej), wywoła `IndexError`. Wycinki (`[a:b]`) są za to bezpieczne — `owoce[2:100]` po prostu zwróci wszystko od indeksu 2 do końca, bez błędu.

<a id="sec3"></a>
## 3. Metody list

Lista jest **mutowalna** — można ją modyfikować w miejscu, bez tworzenia nowej. Najczęściej
używane metody:

| Metoda | Działanie |
|---|---|
| `.append(x)` | dodaje `x` na koniec |
| `.insert(i, x)` | wstawia `x` na pozycję `i` |
| `.remove(x)` | usuwa pierwsze wystąpienie wartości `x` |
| `.pop(i)` | usuwa i **zwraca** element o indeksie `i` (domyślnie ostatni) |
| `.sort()` | sortuje listę w miejscu |
| `.reverse()` | odwraca kolejność w miejscu |
| `in` | sprawdza przynależność, np. `"banan" in owoce` |

````python
zakupy = ["chleb", "mleko", "masło"]

zakupy.append("jajka")
print(zakupy)

zakupy.insert(1, "ser")
print(zakupy)

zakupy.remove("masło")
print(zakupy)

ostatni = zakupy.pop()
print(f"Usunięto: {ostatni}, zostało: {zakupy}")

zakupy.sort()
print(zakupy)

print("mleko" in zakupy)
````

<a id="sec4"></a>
## 4. Iterowanie po liście i `enumerate()`

`for` po liście działa tak samo jak po stringu z modułu 2. Gdy oprócz wartości potrzebny
jest też jej indeks, zamiast ręcznie liczyć licznikiem użyj `enumerate()`.

````python
zakupy = ["chleb", "ser", "jajka"]

for produkt in zakupy:
    print(produkt)

print("---")

for indeks, produkt in enumerate(zakupy):
    print(f"{indeks + 1}. {produkt}")
````

<a id="sec5"></a>
## 5. List comprehension — krótki podgląd

To zwięzły sposób tworzenia nowej listy na podstawie innej sekwencji, w jednej linii.
Ogólna postać: `[wyrażenie for element in sekwencja if warunek]` (warunek jest
opcjonalny). To nie jest nowa "magiczna" konstrukcja — to skrót na pętlę `for` z
`append()`, który poznałeś/aś w poprzedniej sekcji.

````python
kwadraty = [x ** 2 for x in range(1, 6)]
print(kwadraty)

# To samo "długim" sposobem, żeby zobaczyć równoważność:
kwadraty_v2 = []
for x in range(1, 6):
    kwadraty_v2.append(x ** 2)
print(kwadraty_v2)

# Z warunkiem: tylko liczby parzyste
parzyste = [x for x in range(1, 11) if x % 2 == 0]
print(parzyste)
````

> 💡 **Ciekawostka**
>
> List comprehension bywa szybsza niż odpowiadająca jej pętla `for` z `.append()`, bo Python optymalizuje ją wewnętrznie. Nie warto jednak ścigać się z czytelnością — jeśli wyrażenie robi się skomplikowane, zwykła pętla `for` bywa lepszym wyborem.

## 6. Krotki (tuple)

Krotka wygląda jak lista, ale zapisujemy ją w nawiasach okrągłych i jest
**niemutowalna** — raz utworzonej nie da się zmienić (nie ma `.append()`, `.remove()`
itd.). Używa się jej tam, gdzie dane mają się nie zmieniać, np. współrzędne punktu,
albo gdy funkcja chce zwrócić kilka wartości naraz.

````python
wspolrzedne = (52.23, 21.01)   # Warszawa: szerokość, długość geograficzna
print(wspolrzedne[0], wspolrzedne[1])

# Rozpakowanie krotki na osobne zmienne:
szerokosc, dlugosc = wspolrzedne
print(f"Szerokość: {szerokosc}, długość: {dlugosc}")

# Próba zmiany wywoła błąd:
# wspolrzedne[0] = 10   # TypeError: 'tuple' object does not support item assignment

# Klasyczny trik z rozpakowaniem - zamiana wartości miejscami bez zmiennej pomocniczej:
a, b = 1, 2
a, b = b, a
print(a, b)
````

<a id="sec7"></a>
## 7. Słowniki (dict)

Słownik przechowuje pary **klucz: wartość** — zamiast szukać elementu po numerze
indeksu (jak w liście), szukasz go po kluczu, który sam wybierasz (najczęściej string).
Od Pythona 3.7 słowniki pamiętają kolejność wstawiania elementów.

````python
oceny = {"Kamil": 5, "Ania": 4, "Tomek": 3}

print(oceny["Kamil"])          # dostęp po kluczu
oceny["Zosia"] = 5              # dodanie nowej pary
oceny["Tomek"] = 4               # nadpisanie istniejącej wartości
print(oceny)

print(oceny.get("Nieznany", "brak takiej osoby"))  # bezpieczny odczyt bez błędu

for imie, ocena in oceny.items():
    print(f"{imie}: {ocena}")

print(list(oceny.keys()))
print(list(oceny.values()))
````

> ⚠️ **`dict["klucz"]` kontra `.get()`**
>
> Odwołanie `oceny["Nieznany"]` do klucza, którego nie ma, wywoła `KeyError` i przerwie program. `.get("Nieznany", wartosc_domyslna)` zwróci bezpiecznie wartość domyślną zamiast się wywalić — warto go używać, gdy nie masz pewności, że klucz istnieje.

<a id="sec8"></a>
## 8. Zbiory (set)

Zbiór (`set`) przechowuje **unikalne** wartości bez ustalonej kolejności — duplikaty
same się usuwają. Przydaje się do usuwania powtórzeń i do operacji znanych z matematyki:
suma, przecięcie, różnica zbiorów.

````python
liczby = [1, 2, 2, 3, 3, 3, 4]
unikalne = set(liczby)
print(unikalne)          # {1, 2, 3, 4} - duplikaty zniknęły

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a | b)   # suma (union)
print(a & b)   # część wspólna (intersection)
print(a - b)   # różnica (elementy z a, których nie ma w b)
````

## 9. Cztery kolekcje w pigułce

| Typ | Zapis | Mutowalny? | Kolejność | Duplikaty | Dostęp |
|---|---|---|---|---|---|
| `list` | `[1, 2, 3]` | tak | tak | tak | po indeksie |
| `tuple` | `(1, 2, 3)` | nie | tak | tak | po indeksie |
| `dict` | `{"a": 1}` | tak | tak (od 3.7) | klucze unikalne | po kluczu |
| `set` | `{1, 2, 3}` | tak | nie | nie (unikalne) | tylko `in` |

## 10. Ciekawostka: mutowalność i pułapka domyślnego argumentu

Mutowalność list i słowników ma swoje konsekwencje, na które warto uważać już teraz —
jeszcze wrócimy do tego przy okazji funkcji w module 4.

````python
# Dwie zmienne mogą wskazywać na TĘ SAMĄ listę w pamięci:
lista_a = [1, 2, 3]
lista_b = lista_a       # to NIE jest kopia — to ta sama lista pod inną nazwą!

lista_b.append(4)
print(lista_a)   # [1, 2, 3, 4] - zmiana przez lista_b widoczna też w lista_a!

# Żeby zrobić prawdziwą kopię:
lista_c = lista_a.copy()   # albo: list(lista_a)
lista_c.append(99)
print(lista_a)   # bez zmian
print(lista_c)
````

> 💡 **Ciekawostka**
>
> To dokładnie ta sama zasada stoi za znanym błędem "mutable default argument" w funkcjach (`def f(lista=[]):`), gdzie domyślna pusta lista jest tworzona **raz**, przy definicji funkcji, i współdzielona między wszystkimi wywołaniami. Zobaczysz to na żywo w module o funkcjach.

> 📝 **Ćwiczenie 3: Średnia ocen ze słownika**
>
> Mając słownik `oceny = {'Kamil': 5, 'Ania': 4, 'Tomek': 3, 'Zosia': 5}`, oblicz średnią wszystkich ocen (bez ręcznego wypisywania wartości - użyj `.values()` i pętli albo `sum()`).

**👉 Przykładowe rozwiązanie** *(najpierw spróbuj samodzielnie!)*

```python
oceny = {"Kamil": 5, "Ania": 4, "Tomek": 3, "Zosia": 5}

suma = 0
for ocena in oceny.values():
    suma += ocena

srednia = suma / len(oceny)
print(f"Średnia ocen: {srednia:.2f}")

# Krócej, tym samym pomysłem:
# srednia = sum(oceny.values()) / len(oceny)
```

> 📝 **Ćwiczenie 4: Zliczanie liter**
>
> Mając string `tekst = 'programowanie w pythonie'`, zbuduj słownik zliczający, ile razy występuje każda litera (pomiń spacje). Podpowiedź: `slownik.get(litera, 0) + 1`.

**👉 Przykładowe rozwiązanie** *(najpierw spróbuj samodzielnie!)*

```python
tekst = "programowanie w pythonie"
licznik = {}

for litera in tekst:
    if litera == " ":
        continue
    licznik[litera] = licznik.get(litera, 0) + 1

print(licznik)
```

Podpowiedź: `.get(litera, 0)` zwraca dotychczasowy licznik danej litery, a jeśli litera pojawia się pierwszy raz - zwraca 0, więc `+ 1` zawsze bezpiecznie zadziała.

> 📝 **Ćwiczenie 5: Zamiana miejscami przez krotki**
>
> Mając trzy zmienne `a, b, c = 1, 2, 3`, zamień ich wartości tak, żeby `a` miało wartość `c`, `b` miało wartość `a`, a `c` miało wartość `b` - w jednej linii, bez zmiennej pomocniczej (wykorzystaj rozpakowanie krotek).

**👉 Przykładowe rozwiązanie** *(najpierw spróbuj samodzielnie!)*

```python
a, b, c = 1, 2, 3
a, b, c = c, a, b
print(a, b, c)   # 3 1 2
```

> 📝 **Ćwiczenie 7: List comprehension**
>
> Za pomocą list comprehension zbuduj listę sześcianów (x**3) liczb od 1 do 10, ale tylko dla liczb podzielnych przez 3.

**👉 Przykładowe rozwiązanie** *(najpierw spróbuj samodzielnie!)*


> 🔥 **Ćwiczenie 8 (wyzwanie): Mini-inwentarz sklepu**
>
> Stwórz słownik `inwentarz`, gdzie kluczem jest nazwa produktu, a wartością - inny słownik z kluczami `cena` i `ilosc`, np. `{'chleb': {'cena': 4.5, 'ilosc': 20}}`. Dodaj co najmniej 3 produkty. Napisz pętlę, która policzy i wypisze łączną wartość całego inwentarza (`cena * ilosc` zsumowane dla wszystkich produktów), zaokrągloną do 2 miejsc po przecinku.

**👉 Przykładowe rozwiązanie** *(najpierw spróbuj samodzielnie!)*

```python
inwentarz = {
    "chleb": {"cena": 4.5, "ilosc": 20},
    "mleko": {"cena": 3.2, "ilosc": 15},
    "jajka": {"cena": 12.0, "ilosc": 8},
}

wartosc_calkowita = 0
for produkt, dane in inwentarz.items():
    wartosc_produktu = dane["cena"] * dane["ilosc"]
    print(f"{produkt}: {wartosc_produktu:.2f} zł")
    wartosc_calkowita += wartosc_produktu

print(f"Łączna wartość inwentarza: {wartosc_calkowita:.2f} zł")
```

Podpowiedź: to słownik zagnieżdżony w słowniku - `dane` w pętli to sam wewnętrzny słownik `{'cena': ..., 'ilosc': ...}`, więc `dane['cena']` i `dane['ilosc']` działają tak samo jak na każdym innym słowniku.
