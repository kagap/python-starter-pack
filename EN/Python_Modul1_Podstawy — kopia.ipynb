## 3. Your first program — `print()`

The `print()` function outputs text (or the result of an operation) to the screen. It is
the most frequently used function for quickly checking "what is happening" in your code.

````python
print("Hello, world!")
````

`print()` accepts **multiple arguments** at once (separated by commas) as well as two
useful parameters:

- `sep` — what to separate subsequent arguments with (default is a space),
- `end` — what to append at the end (default is a newline character `\n`).

<div style="background-color:#eef6ff;border-left:5px solid #2b6cb0;padding:10px 15px;margin:12px 0;border-radius:4px;color:#1a365d;">
<b>💡 Fun fact</b><br>
In Python 2, <code>print</code> was a statement (<code>print "text"</code>, without parentheses). In Python 3, it became a regular function — one of the most visible differences between the two versions.
</div>

````python
print("Kamil", "PROBS", 2026, sep=" | ")
print("This line", end=" ")
print("ends without a newline, so it appears right here.")
````

## 4. Variables and data types

You create a variable simply by assigning a value to it — **you do not declare the type in
advance**, Python figures it out automatically (*dynamic typing*).

Four basic types to start with:

| Type | Example | Description |
|---|---|---|---|
| `int` | `42` | integer number |
| `float` | `3.14` | floating-point number (with a decimal part) |
| `str` | `"text"` | string of characters |
| `bool` | `True` / `False` | boolean value |

You can check the current type of a variable using the `type()` function.

````python
age = 30
pi = 3.14
name = "Kamil"
is_studying = False

print(age, type(age))
print(pi, type(pi))
print(name, type(name))
print(is_studying, type(is_studying))
````

<div style="background-color:#fff5f5;border-left:5px solid #e53e3e;padding:10px 15px;margin:12px 0;border-radius:4px;color:#742a2a;">
<b>⚠️ Variable naming rules</b><br>
A variable name: must start with a letter or an underscore (not a digit), can contain letters, numbers, and <code>_</code>, is case-sensitive (<code>age</code> and <code>Age</code> are two different variables), and cannot be a Python keyword (e.g., <code>class</code>, <code>for</code>, <code>if</code>). The convention in Python is <code>snake_case</code>, e.g., <code>order_count</code>.
</div>

<div style="background-color:#eef6ff;border-left:5px solid #2b6cb0;padding:10px 15px;margin:12px 0;border-radius:4px;color:#1a365d;">
<b>💡 Fun fact</b><br>
Python uses <i>duck typing</i> — &laquo;if it walks like a duck and quacks like a duck, it's a duck&raquo;. What matters is what a value <b>can do</b>, not its formally declared type. Therefore, the same variable can &laquo;change type&raquo; during execution simply by assigning a new value to it.
</div>

<a id="sec5"></a>
## 5. Comments

A comment is a fragment of code that Python **ignores** during execution — it is meant only
for humans to understand what is happening (or why). Short comments are written after
`#`, and longer ones inside triple quotes.

````python
# This is a single-line comment — Python will ignore it
salary = 8500  # comments can also be placed at the end of a line of code

"""
This is a multi-line comment (or docstring),
useful for describing what an entire block of code
or a function does.
"""
print(salary)
````

<a id="sec6"></a>
## 6. Operators

**Arithmetic**

| Operator | Operation | Example | Result |
|---|---|---|---|
| `+` | addition | `5 + 2` | `7` |
| `-` | subtraction | `5 - 2` | `3` |
| `*` | multiplication | `5 * 2` | `10` |
| `/` | division (always returns `float`) | `5 / 2` | `2.5` |
| `//` | integer division (floor division) | `5 // 2` | `2` |
| `%` | remainder (modulo) | `5 % 2` | `1` |
| `**` | exponentiation | `5 ** 2` | `25` |

**Comparisons** (return `bool`): `==`, `!=`, `<`, `>`, `<=`, `>=`

**Logical**: `and`, `or`, `not`

````python
a = 17
b = 5

print(a + b, a - b, a * b)
print(a / b)    # always float
print(a // b)   # how many full 5s fit into 17
print(a % b)    # remainder

print(a > b, a == b, a != b)
print(a > 10 and b < 10)
print(a > 100 or b < 10)
````

<div style="background-color:#eef6ff;border-left:5px solid #2b6cb0;padding:10px 15px;margin:12px 0;border-radius:4px;color:#1a365d;">
<b>💡 Fun fact</b><br>
The combination of <code>//</code> and <code>%</code> is a classic way to break down numbers, such as converting seconds into hours/minutes/seconds, or checking if a number is even (<code>n % 2 == 0</code>). We will return to this in the conditional statements module.
</div>

## 7. Type conversion and `input()`

The `input()` function retrieves text typed by the user in the console — and **always** returns
it as a `str`, even if the user enters a number. Therefore, the result must be explicitly
converted using `int()` or `float()` if you want to perform math on it.

<div style="background-color:#fff5f5;border-left:5px solid #e53e3e;padding:10px 15px;margin:12px 0;border-radius:4px;color:#742a2a;">
<b>⚠️ Classic beginner mistake</b><br>
<code>input("Enter age: ") + 1</code> will raise a <code>TypeError</code> because you are trying to add a number to text. You need to convert it first: <code>int(input("Enter age: ")) + 1</code>.
</div>

````python
# Uncomment and run to test interactively:
# text = input("Enter your age: ")
# print(type(text))          # always <class 'str'>
# age = int(text)
# print(type(age), age + 1) # now you can do math

# Conversions the other way around also work:
number = 42
print(str(number) + " is now text")
print(float("3.5") + 1)
print(int("10") + int("5"))
````

<a id="sec8"></a>
## 8. Text formatting — f-strings

The most convenient way to insert variable values into text is the **f-string**:
prefix the string literal with the letter `f`, and put variables inside `{curly braces}`.
You can also format numbers right away, e.g., rounding to a specific number of decimal places.

````python
name = "Kamil"
balance = 1234.5678

print(f"Hello, {name}!")
print(f"Your balance is {balance:.2f} PLN")   # rounding to 2 decimal places
print(f"{name} has {balance:,.2f} PLN in account".replace(",", " "))
````

<div style="background-color:#eef6ff;border-left:5px solid #2b6cb0;padding:10px 15px;margin:12px 0;border-radius:4px;color:#1a365d;">
<b>💡 Fun fact</b><br>
F-strings were introduced in Python 3.6 (2016). Previously, you had to use <code>"Hello, {}!".format(name)</code> or the older <code>"Hello, %s!" % name</code>. F-strings are typically the fastest and most readable option, making them the default choice today.
</div>
