## 1. Why functions — the DRY principle

If, back in Module 1, you needed to compute BMI for three different people, you'd have
to copy the same formula three times. That's a quick path to bugs — when something needs
fixing, it's easy to forget one of the copies. Hence the **DRY** principle
(*Don't Repeat Yourself*): repeatable logic gets wrapped in a function and called as many
times as needed.

<a id="sec2"></a>
## 2. `def`, parameters, `return`

A function is defined with the `def` keyword, a name, a list of parameters in
parentheses, and a colon. `return` ends the function's execution and sends a value back
to wherever it was called — a function with no `return` (or a bare `return` with no
value) returns `None`.

````python
def greet(name):
    return f"Hi, {name}!"

message = greet("Kamil")
print(message)
print(greet("Ania"))   # the same function can be called as many times as you like


def calculate_bmi(weight, height):
    return weight / height ** 2

print(calculate_bmi(78, 1.80))
print(calculate_bmi(65, 1.65))
````

> ⚠️ **`return` vs `print`**
>
> A frequent source of confusion for beginners: `print()` only **displays** a value on screen, while `return` **sends it back** for further use in the program (e.g. assigning it to a variable). A function that only `print()`s its result can't have that result reused afterward - `result = calculate_bmi(...)` only works if the function actually uses `return`.

<a id="sec3"></a>
## 3. Default parameter values

A parameter can have a default value, used when the caller doesn't provide one. This
makes a function flexible while still having sensible "out of the box" behavior.

````python
def greet(name, greeting="Hi"):
    return f"{greeting}, {name}!"

print(greet("Kamil"))                    # uses the default "Hi"
print(greet("Kamil", "Welcome"))         # overrides the default
print(greet("Kamil", greeting="Hey"))    # same, but with an explicitly named argument
````

<a id="sec4"></a>
## 4. Positional and keyword arguments

Arguments can be passed **positionally** (in parameter order) or **by name**
(`parameter=value`, where order no longer matters). You can mix the two, but positional
arguments must always come before keyword ones.

````python
def describe_product(name, price, currency="USD"):
    return f"{name}: {price} {currency}"

print(describe_product("Bread", 4.5))
print(describe_product(price=12.0, name="Cheese"))          # keyword order doesn't matter
print(describe_product("Coffee", 15.0, currency="EUR"))

# This raises an error - a positional argument after a keyword one:
# describe_product(name="Coffee", 15.0)
````

<a id="sec5"></a>
## 5. `*args` and `**kwargs`

Sometimes you don't know upfront how many arguments a function will get. `*args`
collects any number of positional arguments into a tuple, and `**kwargs` collects any
number of keyword arguments into a dictionary. The names `args`/`kwargs` are just
convention — what matters is the `*` and `**`.

````python
def sum_everything(*numbers):
    print(numbers, type(numbers))   # this is a tuple
    return sum(numbers)

print(sum_everything(1, 2, 3))
print(sum_everything(1, 2, 3, 4, 5))   # any number of arguments


def print_profile(**data):
    print(data, type(data))         # this is a dictionary
    for key, value in data.items():
        print(f"{key}: {value}")

print_profile(name="Kamil", age=30, city="Warsaw")
````

<a id="sec6"></a>
## 6. Variable scope

A variable defined **inside** a function is local — it only exists while the function
runs, and is invisible outside it. A variable defined **outside** a function (global) is
visible for reading inside the function, but to **change** it from inside the function,
you need to explicitly use the `global` keyword.

````python
visit_count = 0   # global variable

def increment_without_global():
    visit_count = 100   # this creates a NEW LOCAL variable, it does not change the global one!
    print(f"Inside the function: {visit_count}")

def increment_with_global():
    global visit_count
    visit_count += 1

increment_without_global()
print(f"After the first function: {visit_count}")   # unchanged, still 0

increment_with_global()
increment_with_global()
print(f"After two calls with global: {visit_count}")   # 2
````

> ⚠️ **Avoid overusing `global`**
>
> Modifying global variables from inside functions makes it harder to track what changes what and where, especially in larger programs. Good practice: a function should receive data through parameters and return results through `return`, and treat `global` as the exception rather than the norm.

<a id="sec7"></a>
## 7. Docstrings — documenting functions

A docstring is a documentation string placed as the first line of a function's body (in
triple quotes) — it describes what the function does, what arguments it takes, and what
it returns. It's accessible via `help()` or the `.__doc__` attribute.

````python
def calculate_bmi(weight, height):
    """
    Calculates the BMI (Body Mass Index).

    Args:
        weight (float): weight in kilograms
        height (float): height in meters

    Returns:
        float: the BMI value
    """
    return weight / height ** 2

help(calculate_bmi)
print(calculate_bmi.__doc__)
````

<a id="sec8"></a>
## 8. Lambda functions

`lambda` creates a small, anonymous (nameless) function in a single line — handy when a
function is only needed briefly, e.g. as an argument to another function (`sorted()`,
`key=`). Syntax: `lambda arguments: expression` (no `return` — the expression's result is
returned automatically).

````python
square = lambda x: x ** 2
print(square(5))          # same as: def square(x): return x ** 2

# A typical use case: sorting by a custom key
people = [("Ania", 25), ("Kamil", 30), ("Tomek", 20)]
sorted_by_age = sorted(people, key=lambda person: person[1])
print(sorted_by_age)
````

> 💡 **When to use lambda**
>
> Lambda only makes sense for very short, one-expression operations. If the logic gets more involved (more than one line, conditions), a regular `def` function with a sensible name will be much more readable.

<a id="sec9"></a>
## 9. Recursion — a short introduction

A **recursive** function calls itself, with a smaller/simpler argument each time, until
it reaches a so-called **base case** that ends the recursion without another call.
Without a base case, you get infinite recursion (and an error).

````python
def factorial(n):
    if n <= 1:            # base case - recursion stops here
        return 1
    return n * factorial(n - 1)   # calls itself with a smaller argument

print(factorial(5))    # 5 * 4 * 3 * 2 * 1 = 120

# The same thing iteratively, for comparison:
def factorial_iterative(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

print(factorial_iterative(5))
````

> 💡 **Fun fact**
>
> Python limits recursion depth to about 1000 calls by default (`sys.getrecursionlimit()`) - going past that raises a `RecursionError`. This is a deliberate safeguard against a missing base case, which would otherwise freeze the program in an endless chain of calls.

    if n <= 1:              # base case

Explanation: this version recomputes the same values over and over (e.g. `fibonacci(3)` gets calculated from scratch every time it's needed) - the number of calls grows exponentially with `n`.
