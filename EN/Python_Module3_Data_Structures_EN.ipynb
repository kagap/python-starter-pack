## 1. Why we need collections

Imagine a shopping list of 15 items. Creating 15 separate variables (`item1`, `item2`,
...) would be absurd — and impossible to handle with a loop, since you can't write
`for i in range(15): item{i}`. That's why Python has built-in **collection** types that
hold many values under a single variable name.

<a id="sec2"></a>
## 2. Lists — creating, indexing, slicing

A list (`list`) is an ordered, **mutable** (changeable after creation) collection of
values, written in square brackets. It can hold elements of different types at once,
though in practice it usually holds a single type.

Elements are numbered from **0**, and negative indices count from the end (`-1` is the
last element).

````python
fruits = ["apple", "banana", "pear", "orange", "kiwi"]

print(fruits[0])       # first element
print(fruits[-1])      # last element
print(fruits[1:3])     # slice: indices 1 and 2 (3 is excluded)
print(fruits[:2])      # from the start up to (not including) index 2
print(fruits[2:])      # from index 2 to the end
print(len(fruits))     # number of elements
````

> ⚠️ **Watch out for IndexError**
>
> Trying to access an index that doesn't exist (e.g. `fruits[10]` on a five-element list) raises an `IndexError`. Slices (`[a:b]`) are safe though — `fruits[2:100]` simply returns everything from index 2 to the end, no error.

<a id="sec3"></a>
## 3. List methods

A list is **mutable** — you can modify it in place without creating a new one. The most
commonly used methods:

| Method | What it does |
|---|---|
| `.append(x)` | adds `x` at the end |
| `.insert(i, x)` | inserts `x` at position `i` |
| `.remove(x)` | removes the first occurrence of value `x` |
| `.pop(i)` | removes and **returns** the item at index `i` (last by default) |
| `.sort()` | sorts the list in place |
| `.reverse()` | reverses the order in place |
| `in` | checks membership, e.g. `"banana" in fruits` |

````python
groceries = ["bread", "milk", "butter"]

groceries.append("eggs")
print(groceries)

groceries.insert(1, "cheese")
print(groceries)

groceries.remove("butter")
print(groceries)

last_item = groceries.pop()
print(f"Removed: {last_item}, left: {groceries}")

groceries.sort()
print(groceries)

print("milk" in groceries)
````

<a id="sec4"></a>
## 4. Iterating a list and `enumerate()`

`for` over a list works the same way as over a string from Module 2. When you need both
the value and its index, don't count manually — use `enumerate()`.

````python
groceries = ["bread", "cheese", "eggs"]

for item in groceries:
    print(item)

print("---")

for index, item in enumerate(groceries):
    print(f"{index + 1}. {item}")
````

<a id="sec5"></a>
## 5. List comprehensions — a quick preview

This is a compact way to build a new list from another sequence, in a single line. The
general form: `[expression for item in sequence if condition]` (the condition is
optional). It's not a brand new "magic" construct — it's shorthand for the `for` loop
with `.append()` you saw in the previous section.

````python
squares = [x ** 2 for x in range(1, 6)]
print(squares)

# The same thing "the long way", to see the equivalence:
squares_v2 = []
for x in range(1, 6):
    squares_v2.append(x ** 2)
print(squares_v2)

# With a condition: only even numbers
evens = [x for x in range(1, 11) if x % 2 == 0]
print(evens)
````

> 💡 **Fun fact**
>
> A list comprehension can be faster than the equivalent `for` loop with `.append()`, since Python optimizes it internally. Still, don't chase performance at the expense of readability — if the expression gets complicated, a plain `for` loop is often the better choice.

## 6. Tuples

A tuple looks like a list, but is written in round brackets and is **immutable** —
once created, it can't be changed (no `.append()`, `.remove()`, etc.). It's used when
data shouldn't change, like the coordinates of a point, or when a function wants to
return several values at once.

````python
coordinates = (52.23, 21.01)   # Warsaw: latitude, longitude
print(coordinates[0], coordinates[1])

# Unpacking a tuple into separate variables:
latitude, longitude = coordinates
print(f"Latitude: {latitude}, longitude: {longitude}")

# Trying to change it raises an error:
# coordinates[0] = 10   # TypeError: 'tuple' object does not support item assignment

# A classic unpacking trick - swap values without a helper variable:
a, b = 1, 2
a, b = b, a
print(a, b)
````

<a id="sec7"></a>
## 7. Dictionaries

A dictionary stores **key: value** pairs — instead of looking something up by an index
number (like a list), you look it up by a key you choose yourself (usually a string).
Since Python 3.7, dictionaries remember the order items were inserted in.

````python
grades = {"Kamil": 5, "Ania": 4, "Tomek": 3}

print(grades["Kamil"])          # access by key
grades["Zosia"] = 5               # add a new pair
grades["Tomek"] = 4                # overwrite an existing value
print(grades)

print(grades.get("Unknown", "no such person"))  # safe read, no error

for name, grade in grades.items():
    print(f"{name}: {grade}")

print(list(grades.keys()))
print(list(grades.values()))
````

> ⚠️ **`dict["key"]` vs `.get()`**
>
> Accessing `grades["Unknown"]` for a key that does not exist raises a `KeyError` and crashes the program. `.get("Unknown", default_value)` safely returns the default instead of blowing up - worth using whenever you're not sure a key exists.

<a id="sec8"></a>
## 8. Sets

A set (`set`) stores **unique** values with no fixed order — duplicates disappear on
their own. It's handy for removing repeats, and for the classic math operations: union,
intersection, difference.

````python
numbers = [1, 2, 2, 3, 3, 3, 4]
unique = set(numbers)
print(unique)          # {1, 2, 3, 4} - duplicates are gone

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a | b)   # union
print(a & b)   # intersection
print(a - b)   # difference (items in a that aren't in b)
````

<a id="sec9"></a>
## 9. The four collections at a glance

| Type | Syntax | Mutable? | Ordered | Duplicates | Access |
|---|---|---|---|---|---|
| `list` | `[1, 2, 3]` | yes | yes | yes | by index |
| `tuple` | `(1, 2, 3)` | no | yes | yes | by index |
| `dict` | `{"a": 1}` | yes | yes (since 3.7) | keys unique | by key |
| `set` | `{1, 2, 3}` | yes | no | no (unique) | `in` only |

## 10. Fun fact: mutability and the default-argument trap

The mutability of lists and dictionaries has consequences worth watching for right
now — we'll come back to this when we get to functions in Module 4.

````python
# Two variables can point to the SAME list in memory:
list_a = [1, 2, 3]
list_b = list_a       # this is NOT a copy - it's the same list under a different name!

list_b.append(4)
print(list_a)   # [1, 2, 3, 4] - the change via list_b shows up in list_a too!

# To make a real copy:
list_c = list_a.copy()   # or: list(list_a)
list_c.append(99)
print(list_a)   # unchanged
print(list_c)
````

> 💡 **Fun fact**
>
> This is exactly the same principle behind the well-known "mutable default argument" bug in functions (`def f(items=[]):`), where the default empty list is created **once**, when the function is defined, and shared across every call. You'll see this live in the functions module.

> 🔥 **Exercise 8 (challenge): Mini store inventory**
>
> Create a dictionary `inventory` where the key is a product name and the value is another dictionary with keys `price` and `qty`, e.g. `{'bread': {'price': 4.5, 'qty': 20}}`. Add at least 3 products. Write a loop that computes and prints the total value of the entire inventory (`price * qty` summed over all products), rounded to 2 decimal places.

**👉 Sample solution** *(try it yourself first!)*

```python
inventory = {
    "bread": {"price": 4.5, "qty": 20},
    "milk": {"price": 3.2, "qty": 15},
    "eggs": {"price": 12.0, "qty": 8},
}

total_value = 0
for product, data in inventory.items():
    product_value = data["price"] * data["qty"]
    print(f"{product}: ${product_value:.2f}")
    total_value += product_value

print(f"Total inventory value: ${total_value:.2f}")
```

Hint: this is a dictionary nested inside a dictionary - `data` in the loop is itself the inner dictionary `{'price': ..., 'qty': ...}`, so `data['price']` and `data['qty']` work exactly like on any other dictionary.
