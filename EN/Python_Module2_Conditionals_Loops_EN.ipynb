## 1. Truth, falsehood, and "truthy / falsy"

In Module 1 you met the `bool` type (`True` / `False`) and comparison operators. Those
logical expressions are exactly what will drive `if` statements and loops.

Here's something that makes life easier: `if` doesn't *literally* require a `bool`
value — every value in Python can be evaluated as "true-ish" (**truthy**) or "false-ish"
(**falsy**).

````python
# Values Python treats as "falsy":
print(bool(0))       # False
print(bool(0.0))     # False
print(bool(""))      # False - empty string
print(bool(None))    # False
print(bool([]))      # False - empty list (we'll cover this properly in module 3)

# Pretty much everything else is "truthy":
print(bool(1))
print(bool(-5))
print(bool("0"))     # WATCH OUT: this is the string "0", not empty -> True!
print(bool("text"))
````

<div style="background-color:#eef6ff;border-left:5px solid #2b6cb0;padding:10px 15px;margin:12px 0;border-radius:4px;color:#1a365d;">
<b>💡 Fun fact</b><br>
The expression <code>bool(&quot;0&quot;)</code> gives <code>True</code>, because it's a non-empty string (the fact that it contains the character &quot;0&quot; doesn't matter). That's a classic trap when validating data from <code>input()</code> — an empty string is falsy, but a string containing the digit &quot;0&quot; is not.
</div>

## 2. The `if` / `elif` / `else` statement

Basic syntax:

```python
if condition:
    # code that runs if the condition is true
elif other_condition:
    # code that runs if the first condition was false, but this one is true
else:
    # code that runs if none of the above conditions were true
```

`elif` and `else` are optional — you can have just `if`, `if`/`else`, or any number of
`elif`s in between. Conditions are checked **in order, top to bottom**, and only the
block for the **first** condition that turns out true runs — the rest are skipped.

````python
age = 20

if age < 13:
    print("Child")
elif age < 18:
    print("Teenager")
elif age < 65:
    print("Adult")
else:
    print("Senior")
````

<div style="background-color:#fff5f5;border-left:5px solid #e53e3e;padding:10px 15px;margin:12px 0;border-radius:4px;color:#742a2a;">
<b>⚠️ A common mistake</b><br>
A single <code>=</code> is assignment (<code>x = 5</code>), while a double <code>==</code> is comparison (<code>x == 5</code>). Mixing them up inside an <code>if</code> immediately raises a syntax error in Python — unlike some other languages, where that mistake can hide for a long time.
</div>

## 3. Indentation — not just cosmetics

This is one of Python's most distinctive features: **indentation is part of the syntax**,
not just a style choice. The block belonging to an `if`, `while`, `for`, a function, etc.
must be consistently indented (usually 4 spaces).

````python
# Correct:
if True:
    print("This is inside the if block")
print("This is no longer inside the if block")

# The following (uncommented) would raise an IndentationError, because it's missing
# indentation:
# if True:
# print("syntax error")
````

<div style="background-color:#fff5f5;border-left:5px solid #e53e3e;padding:10px 15px;margin:12px 0;border-radius:4px;color:#742a2a;">
<b>⚠️ Watch out for mixing spaces and tabs</b><br>
Python 3 doesn't allow mixing spaces and tabs within the same block — it will raise a <code>TabError</code>. In Jupyter, the Tab key usually inserts 4 spaces on its own, so in practice this rarely comes up, but it's good to know where a confusing error might come from when pasting code from different sources.
</div>

## 4. Nested conditions and the conditional operator

An `if` can contain another `if` inside it (nesting) — useful when a second decision only
makes sense once the first condition is already true. For simple "value A or B depending
on a condition" cases, there's also a shorter form — the **conditional (ternary)
operator**: `value_a if condition else value_b`.

````python
temperature = 5
is_raining = True

if temperature < 0:
    print("Dress warmly")
    if is_raining:
        print("...and bring an umbrella, though that's rare in freezing weather")
else:
    print("It's not cold")

# Same idea as a simple if/else, written on one line:
status = "warm" if temperature > 15 else "cold"
print(status)
````

## 5. The `while` loop

A `while` loop runs a block of code **as long as a condition is true** — it checks the
condition before every repetition. It's on you to make sure the condition eventually
becomes false (otherwise you get an infinite loop).

````python
counter = 1
while counter <= 5:
    print(f"Pass number {counter}")
    counter += 1   # same as: counter = counter + 1

print("Loop finished")
````

<div style="background-color:#fff5f5;border-left:5px solid #e53e3e;padding:10px 15px;margin:12px 0;border-radius:4px;color:#742a2a;">
<b>⚠️ Infinite loops</b><br>
If you forget to update the controlling variable (e.g. <code>counter += 1</code>), the condition never changes and the loop runs forever, freezing the Jupyter kernel. If that happens, stop execution with the <b>■ (Stop)</b> button in the toolbar, or via <i>Kernel &rarr; Interrupt</i> from the menu.
</div>

<div style="background-color:#eef6ff;border-left:5px solid #2b6cb0;padding:10px 15px;margin:12px 0;border-radius:4px;color:#1a365d;">
<b>💡 Pattern: while True + break</b><br>
A very common pattern is <code>while True:</code> (a condition that's always true), with an exit condition inside the loop via <code>break</code> — handy for things like validating user input, where you want to keep asking &quot;until the answer actually makes sense&quot;. You'll see this in the exercises.
</div>

<a id="sec6"></a>
## 6. `break` and `continue`

Two statements for controlling a loop from the inside:

- **`break`** — immediately **exits the whole loop** (stops it completely).
- **`continue`** — **skips the rest of the current pass** and moves on to the next one.

````python
# break - find the first number divisible by 7 in the range 1-100
n = 1
while n <= 100:
    if n % 7 == 0:
        print(f"Found: {n}")
        break
    n += 1

print("---")

# continue - print numbers 1-10, but skip the even ones
i = 0
while i < 10:
    i += 1
    if i % 2 == 0:
        continue
    print(i)
````

## 7. The `for` loop and `range()`

A `for` loop iterates over a **sequence** — most often the result of the `range()`
function, which generates a sequence of whole numbers. `range()` has three call forms:

| Call | What it generates |
|---|---|
| `range(stop)` | `0, 1, 2, ..., stop-1` |
| `range(start, stop)` | `start, start+1, ..., stop-1` |
| `range(start, stop, step)` | every `step`, from `start` to `stop-1` (step can be negative) |

Notice: **`stop` is never reached** — `range(5)` ends at 4.

````python
for i in range(5):
    print(i, end=" ")
print()

for i in range(2, 10, 2):
    print(i, end=" ")
print()

for i in range(10, 0, -1):     # counting down
    print(i, end=" ")
print("\nGo!")
````

<div style="background-color:#eef6ff;border-left:5px solid #2b6cb0;padding:10px 15px;margin:12px 0;border-radius:4px;color:#1a365d;">
<b>💡 `for` vs `while`</b><br>
If you know upfront <b>how many times</b> something should repeat (or you're iterating over a known sequence) — use <code>for</code>. If the number of repetitions depends on a condition you'll only know during execution (e.g. &quot;until the user guesses correctly&quot;) — use <code>while</code>. Any <code>for</code> loop can technically be written as a <code>while</code>, but it's rarely worth it.
</div>

<a id="sec8"></a>
## 8. Iterating over text (and a preview of lists)

`for` works on any sequence, not just `range()`. A string is also a sequence of
characters, so you can iterate over it directly.

````python
for letter in "Python":
    print(letter)

# Let's count the vowels in a word:
word = "programming"
vowels = "aeiou"
vowel_count = 0
for letter in word:
    if letter in vowels:
        vowel_count += 1

print(f"Number of vowels in '{word}': {vowel_count}")
````

## 9. Nested loops

A loop can contain another loop inside it — then for **every** pass of the outer loop,
the **entire** inner loop runs. Classic example: a multiplication table.

````python
for a in range(1, 4):
    for b in range(1, 4):
        print(f"{a} x {b} = {a * b}")
    print("---")   # separator after each "round" of the outer loop
````

<div style="background-color:#fff5f5;border-left:5px solid #e53e3e;padding:10px 15px;margin:12px 0;border-radius:4px;color:#742a2a;">
<b>⚠️ Watch the complexity</b><br>
Two nested loops of 100 passes each is already 10,000 executions of the inner block (100 x 100). With larger ranges, nested loops can noticeably slow a program down — a topic we'll return to when we talk about computational complexity.
</div>

## 10. Fun facts: `for...else`, `pass`

Two less common, but worth-knowing constructs.

````python
# for...else: the "else" block runs ONLY if the loop finished normally
# (without a break). Handy when searching for something.
numbers = [4, 8, 15, 16, 23, 42]
target = 99

for n in numbers:
    if n == target:
        print("Found it!")
        break
else:
    print(f"{target} is not in the list")

# pass: "do nothing" - a placeholder when the syntax requires a block,
# but you don't have anything to put in it yet.
if True:
    pass  # I'll fill this in later
print("The program runs fine even with an empty if block")
````

<div style="background-color:#eef6ff;border-left:5px solid #2b6cb0;padding:10px 15px;margin:12px 0;border-radius:4px;color:#1a365d;">
<b>💡 Fun fact</b><br>
<code>for...else</code> tends to confuse beginners (and not only them) — the word &quot;else&quot; suggests &quot;otherwise&quot;, but here it really means &quot;if there was no <code>break</code>&quot;. Guido van Rossum has admitted that if he designed it again, he'd probably name this construct something else.
</div>

Write a game: the computer picks a random number between 1 and 100 (<code>import random</code>, <code>random.randint(1, 100)</code>), and the user guesses via <code>input()</code>. ```
