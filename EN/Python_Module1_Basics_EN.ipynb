## 2. What is Python and why bother?

Python is a programming language built around **code readability** — its syntax reads
close to natural English, and indentation isn't just cosmetic, it's part of the syntax
(more on that in the module about conditionals and loops).

It's used everywhere: data analysis and AI/ML (which is likely part of why you ended up
in this notebook), automation, web development (Django, Flask), scripting, testing, and
even robotics.

## 3. Your first program — `print()`

The `print()` function outputs text (or the result of a computation) to the screen. It's
the most commonly used function for quickly checking "what's going on" in your code.

````python
print("Hello, world!")
````

`print()` can take **multiple arguments** at once (separated by commas), plus two
handy parameters:

- `sep` — what to put between the arguments (a space by default),
- `end` — what to add at the very end (a newline `\n` by default).

<div style="background-color:#eef6ff;border-left:5px solid #2b6cb0;padding:10px 15px;margin:12px 0;border-radius:4px;color:#1a365d;">
<b>💡 Fun fact</b><br>
In Python 2, <code>print</code> was a statement (<code>print "text"</code>, no parentheses). In Python 3 it became a regular function — one of the most visible differences between the two versions of the language.
</div>

````python
print("Kamil", "PROBS", 2026, sep=" | ")
print("This line", end=" ")
print("ends without a newline, so it continues right here.")
````

## 4. Variables and data types

You create a variable simply by assigning a value to it — **you don't declare a type up
front**, Python figures it out on its own (so-called *dynamic typing*).

Four basic types to start with:

| Type | Example | Description |
|---|---|---|
| `int` | `42` | whole number |
| `float` | `3.14` | decimal (floating-point) number |
| `str` | `"text"` | a string of characters |
| `bool` | `True` / `False` | a logical (boolean) value |

You can check the type of a variable's current value with the `type()` function.

````python
age = 30
pi = 3.14
name = "Kamil"
is_studying = False

print(age, type(age))
print(pi, type(pi))
print(name, type(name))
print(is_studying, type(is_studying))
````

<div style="background-color:#fff5f5;border-left:5px solid #e53e3e;padding:10px 15px;margin:12px 0;border-radius:4px;color:#742a2a;">
<b>⚠️ Naming variables</b><br>
A variable name: starts with a letter or underscore (never a digit), can contain letters, digits and <code>_</code>, is case-sensitive (<code>age</code> and <code>Age</code> are two different variables), and can't be a Python keyword (e.g. <code>class</code>, <code>for</code>, <code>if</code>). The convention in Python is <code>snake_case</code>, e.g. <code>order_count</code>.
</div>

<div style="background-color:#eef6ff;border-left:5px solid #2b6cb0;padding:10px 15px;margin:12px 0;border-radius:4px;color:#1a365d;">
<b>💡 Fun fact</b><br>
Python uses so-called <i>duck typing</i> — &laquo;if it walks like a duck and quacks like a duck, it's a duck&raquo;. What matters is what a value <b>can do</b>, not its formally declared type. That's why the same variable can effectively &laquo;change type&raquo; during a program's run — simply by assigning it a new value.
</div>

<a id="sec5"></a>
## 5. Comments

A comment is a piece of code that Python **ignores** when running — it's there purely for
humans, so they understand what's happening (or why). Short comments start with `#`,
longer ones go in triple quotes.

````python
# This is a single-line comment — Python will ignore it
salary = 8500  # a comment can also go at the end of a line of code

"""
This is a longer comment (also called a docstring),
handy for e.g. describing what a whole block of code
or a function does.
"""
print(salary)
````

<a id="sec6"></a>
## 6. Operators

**Arithmetic**

| Operator | Operation | Example | Result |
|---|---|---|---|
| `+` | addition | `5 + 2` | `7` |
| `-` | subtraction | `5 - 2` | `3` |
| `*` | multiplication | `5 * 2` | `10` |
| `/` | division (always returns `float`) | `5 / 2` | `2.5` |
| `//` | floor (integer) division | `5 // 2` | `2` |
| `%` | remainder (modulo) | `5 % 2` | `1` |
| `**` | exponentiation | `5 ** 2` | `25` |

**Comparisons** (return `bool`): `==`, `!=`, `<`, `>`, `<=`, `>=`

**Logical**: `and`, `or`, `not`

````python
a = 17
b = 5

print(a + b, a - b, a * b)
print(a / b)    # always a float
print(a // b)   # how many full "fives" fit into 17
print(a % b)    # what's left over

print(a > b, a == b, a != b)
print(a > 10 and b < 10)
print(a > 100 or b < 10)
````

<div style="background-color:#eef6ff;border-left:5px solid #2b6cb0;padding:10px 15px;margin:12px 0;border-radius:4px;color:#1a365d;">
<b>💡 Fun fact</b><br>
Combining <code>//</code> and <code>%</code> is the classic way to break a number down into parts — e.g. seconds into hours/minutes/seconds, or checking whether a number is even (<code>n % 2 == 0</code>). We'll come back to this when we cover conditionals in the next module.
</div>

## 7. Type conversion and `input()`

The `input()` function reads text typed by the user in the console — and **always**
returns it as a `str`, even if the person types a number. So you need to deliberately
convert the result with `int()` or `float()` before doing any math with it.

<div style="background-color:#fff5f5;border-left:5px solid #e53e3e;padding:10px 15px;margin:12px 0;border-radius:4px;color:#742a2a;">
<b>⚠️ A classic beginner mistake</b><br>
<code>input('Enter your age: ') + 1</code> will crash (<code>TypeError</code>), because you're trying to add a number to text. You need to convert first: <code>int(input('Enter your age: ')) + 1</code>.
</div>

````python
# Uncomment and run to try it interactively:
# text = input("Enter your age: ")
# print(type(text))          # always <class 'str'>
# age = int(text)
# print(type(age), age + 1)  # now you can do math with it

# Conversions the other way work too:
number = 42
print(str(number) + " is now text")
print(float("3.5") + 1)
print(int("10") + int("5"))
````

<a id="sec8"></a>
## 8. Formatting text — f-strings

The most convenient way to insert variable values into text is an **f-string**: put an
`f` right before the quote, and put variables inside `{curly braces}`. You can also
format numbers right there — e.g. rounding to a specific number of decimal places.

````python
name = "Kamil"
balance = 1234.5678

print(f"Hi, {name}!")
print(f"Your balance is ${balance:.2f}")   # rounded to 2 decimal places
print(f"{name} has ${balance:,.2f} in the account")
````

<div style="background-color:#eef6ff;border-left:5px solid #2b6cb0;padding:10px 15px;margin:12px 0;border-radius:4px;color:#1a365d;">
<b>💡 Fun fact</b><br>
F-strings only arrived in Python 3.6 (2016). Before that, people used <code>"Hi, {}!".format(name)</code> or the even older <code>"Hi, %s!" % name</code>. F-strings are usually the fastest and most readable option, which is why they're the default choice today.
</div>
